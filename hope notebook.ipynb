```python
# Import libaries
import pandas as pd
import cbsodata

data = pd.DataFrame(cbsodata.get_data("85618NED"))
print("Afmetingen:", data.shape)
print(data.columns.tolist())
display(data.head())
```

```python
# Dataframe maken
data["SoortRegio_2"] = data["SoortRegio_2"].str.strip() #Verwijdert de extra spaties voor en achter de tekst

buurten = data[data["SoortRegio_2"] == "Buurt"].copy()

print("Afmetingen buurten:", buurten.shape)
buurten.head()
```

```python
print(data["SoortRegio_2"].value_counts(dropna = False))

print("Afmetingen buurten:", buurten.shape)
buurten.head()
```

```python
#Bekijk hoeveel ontbrekende waarden elke kolom heeft
ontbrekend = buurten.isnull().sum() #telt hoeveel waarden ontbreken per kolom

print(ontbrekend[ontbrekend > 0].sort_values(ascending = False)) 
# Selecteert alleen kolommen met ontrbekende waarden en zet de kolommen met de meest ontbrekende waarde bovenaan
```

```python
#Ontbrekende waardes vinden en daarvan de percentage per categorie. 13333 ontbrekende waardes is namelijk veel meer dan bijv 11, beide hebben verschillende aanpak
ontbrekend_pct = (buurten.isnull().mean() * 100).round(2)

overzicht_ontbrekend = pd.DataFrame({
    "Ontbrekende waarden": buurten.isnull().sum(),
    "Percentage ontbrekend": ontbrekend_pct
})

overzicht_ontbrekend = overzicht_ontbrekend[
    overzicht_ontbrekend["Ontbrekende waarden"]> 0 
].sort_values("Percentage ontbrekend", ascending=False)

display(overzicht_ontbrekend)
```

## Kolommen met ontbrekende waarden
Naast de relevantie van variabelen bekijken we hoeveel waarden ontbreken. Kolommen met veel ontbrekende waarden kunnen minder bruikbaar zijn voor het model. We kiezen hiervoor een vaste grens en beoordleen ook of de betreffende kolommen inhoudelijk belangrijk is. Ontbrekende waarden worden niet automatisch verwijderd wanneer de variabele belangrijke informatie bevat.

#### Onderbouwing te veel missende waardes
We hebben om te beginnen met een grens van 50%. Waarom? Als meer dan de helft van de waarde ontbreekt, bevat een kolom relatief weinig bruikbare informatie. Het verwijderen van zo'n kolom kan de analuse overzichtelijker maken. Een belangrijke variabele verwijderen we niet zomaar alleen vanwege ontrebeknde waardem we beoordelen ook de inhoud.

```python
# Verwijderen van kolommen met te veel missende waardes
niet_relevant = [
    "ID",
    'SoortRegio_2',
    'Codering_3',
    'IndelingswijzigingGemeenteWijkBuurt_4',
    'GeboorteTotaal_25',
    'GeboorteRelatief_26',
    'SterfteTotaal_27',
    'SterfteRelatief_28',
    'BedrijfsvestigingenTotaal_100',
    'ALandbouwBosbouwEnVisserij_101',
    'BFNijverheidEnEnergie_102',
    'GIHandelEnHoreca_103',
    'HJVervoerInformatieEnCommunicatie_104',
    'KLFinancieleDienstenOnroerendGoed_105',
    'MNZakelijkeDienstverlening_106',
    'OQOverheidOnderwijsEnZorg_107',
    'RUCultuurRecreatieOverigeDiensten_108',
    'PersonenautoSTotaal_109',
    'PersonenautoSBrandstofBenzine_110',
    'PersonenautoSOverigeBrandstof_111',
    'PersonenautoSPerHuishouden_112',
    'PersonenautoSNaarOppervlakte_113',
    'Motorfietsen_114',
    'AfstandTotHuisarsenpraktijk_115',
    'AfstandTotGroteSupermarkt_116',
    'AfstandTotKinderdagverblijf_117',
    'AfstandTotSchool_118',
    'ScholenBinnen3km_119',
    'OppervlakteWater_122',
    'MeestVoorkomendePostcode_123',
    'Dekkingspercentage_124'
]

#Alleen kolommen verwijderen die daadwerlijkk bestaan
niet_relevant = [
    kolom for kolom in niet_relevant
    if kolom in buurten.columns
]
print("Aantal kolommen voor: ", buurten.shape[1])

buurten_schoon = buurten.drop(columns = niet_relevant)

print("Aantal kolommen na", buurten_schoon.shape[1])
```

```python
## Verwijder kolommen met meer dan 50% ontrbekende waarden
# percentage ontbrekende waarde per kolom
percentage_nan = buurten_schoon.isnull().mean() * 100

#kolommem met meer dan 50 procent ontbrekende waarden
te_veel_nan = percentage_nan[percentage_nan > 50 ].index.tolist()

print("Kolommen verwijderd vanwege ontrbekende waarde:")
print(te_veel_nan)

#verwijder de kolommen
buurten_schoon = buurten_schoon.drop(columns=te_veel_nan)
print("Aantal kolommen na opschonen:", buurten_schoon.shape[1])

```

```python
print("Totaal ontbrekende waarden:",
buurten_schoon.isnull().sum().sum())

print('Kolommen met ontbrekende waarden:',
buurten_schoon.isnull().sum()[buurten_schoon.isnull().sum() > 0])
```

## Verwijderen van kolommen met ontbrekende waarden
Wij hebben ervoor gekozen om kolommen met meer dan 50% ontbrekende waarden te verwijderen. Volgens Veridical Data Science is het verwijderen van variabelen met bijvoorbeeld meer dan 50% ontrbekende waarde een mogelijke aanpak tijdens de datavoorbereiding. De auteurs benadrukken dat deze grens een bewuste keuze is en afhankelijk is van het onderzoek (Veridical Data Science, z.d.).

Ook het CBS geeft aan dat er geen universele grens bestaat voor het aantal ontbrekende waarden dat aanvaardbaar is. Dit hangt onder andere af van de statistische analyse die wordt uitgevoerd. (Centraal Bureau voor de Statistiek, z.d.).

Voor ons onderzoek hebben we de grens van 50% gekozen om kolommen te verwijderen waarin meer dan de helft van de waarden ontbreekt. Hierdoor beperken we het gebruik van variabelen met veel ontbrekende informatie. We controleren daarbij wel of belangrijke variabelen voor het voorspellen van het elektriciteitsverbruik behouden blijven


```python
# Een grens stellen, drempel op 50%
drempel = 0.5 * len(buurten_schoon)

#selecteren kolommen met minder dan 50% ontbrekende waarden
relevante_kolommen = buurten_schoon.columns[
    buurten_schoon.isna().sum() < drempel
]

#Nieuwe dataframe maken met alleen relevante kolommen
buurten_schoon = buurten[relevante_kolommen].copy()
```

#### Basisstatistieken

```python
#Totalen per gemeente bereken
totalen_gemeente = buurten_schoon.groupby("Gemeentenaam_1").agg({
    "AantalInwoners_5": "sum",
    "Mannen_6": "sum",
    "Vrouwen_7": "sum"
})
display(totalen_gemeente.head(10))
```

```python
# Basisstatistieken:
# Percentages mannen per buurt
buurten_schoon["Percentage_mannen"] = (
    buurten_schoon["Mannen_6"] /
    buurten_schoon["AantalInwoners_5"] * 100
)
# Percentages vrouwen per buurt
buurten_schoon ["Percentage_vrouwen"] = (
    buurten_schoon["Vrouwen_7"] / 
    buurten_schoon["AantalInwoners_5"] * 100
)

#resultaten bekijken
display(
    buurten_schoon[
        ["WijkenEnBuurten",
        "Gemeentenaam_1",
        "AantalInwoners_5",
        "Percentage_mannen",
        "Percentage_vrouwen"
        ]
    ].head(10).round(2)
)
```

```python
# Bereken het verschil tussen de aantallen
verschil = (
    buurten_schoon["Mannen_6"]
    + buurten_schoon["Vrouwen_7"]
    - buurten_schoon["AantalInwoners_5"]
)

# Bekijk hoe groot de verschillen zijn
print("Aantal buurten met verschil:", (verschil != 0).sum())
print("Verdeling van de verschillen:")
display(verschil.value_counts().sort_index())
```
